```python
# Question 9: Character frequency in "hello"
print("\nQuestion 9: Count frequency of each character in string 'hello'")
word = "hello"
frequency = {}
for char in word:
    frequency[char] = frequency.get(char, 0) + 1
print(frequency)
```
