```python
# Question 10: Prime numbers between 1 and 50
print("\nQuestion 10: Create a list of prime numbers between 1 and 50")
primes = []
for n in range(2, 51):
    if all(n % d != 0 for d in range(2, int(n**0.5) + 1)):
        primes.append(n)
print(primes)
```

```python
# Question 33: Create a nested list
print("\nQuestion 33: Create a nested list with different levels: [1, [2, 3], [4, [5, 6]], 7]")
nested = [1, [2, 3], [4, [5, 6]], 7]
print(nested)
```

```python
# Question 34: Count depth of nesting
print("\nQuestion 34: Count the depth of nesting in [1, [2, 3], [4, [5, 6]], 7]")
def nesting_depth(items):
    if not isinstance(items, list):
        return 0
    if not items:
        return 1
    return 1 + max(
        (nesting_depth(item) for item in items
         if isinstance(item, list)),
        default=0
    )

print(nesting_depth(nested))
```
