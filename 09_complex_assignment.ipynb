```python
# Question 1: Create complex number 5 + 3j
print("Question 1: Create complex number 5 + 3j")
z1 = 5 + 3j
print(z1)
```

```python
# Question 2: Find the real part of 7 - 2j
print("\nQuestion 2: Find the real part of complex number 7 - 2j")
z2 = 7 - 2j
print(z2.real)
```

```python
# Question 3: Find the imaginary part of 4 + 6j
print("\nQuestion 3: Find the imaginary part of complex number 4 + 6j")
z3 = 4 + 6j
print(z3.imag)
```

```python
# Question 4: Add two complex numbers
print("\nQuestion 4: Add two complex numbers: (3 + 2j) and (1 + 4j)")
z4 = (3 + 2j) + (1 + 4j)
print(z4)
```

```python
# Question 5: Multiply two complex numbers
print("\nQuestion 5: Multiply two complex numbers: (2 + 3j) and (1 + 2j)")
z5 = (2 + 3j) * (1 + 2j)
print(z5)
```

```python
# Question 6: Find the magnitude of 6 + 8j
print("\nQuestion 6: Find the magnitude of complex number 6 + 8j")
z6 = 6 + 8j
print(abs(z6))
```

```python
# Question 7: Find the conjugate of 5 - 7j
print("\nQuestion 7: Find the conjugate of complex number 5 - 7j")
z7 = 5 - 7j
print(z7.conjugate())
```

```python
# Question 8: Subtract complex numbers
print("\nQuestion 8: Subtract complex numbers: (10 + 5j) - (3 + 2j)")
z8 = (10 + 5j) - (3 + 2j)
print(z8)
```

```python
# Question 9: Divide complex numbers
print("\nQuestion 9: Divide complex numbers: (8 + 6j) / (2 + 1j)")
z9 = (8 + 6j) / (2 + 1j)
print(z9)
```

```python
# Question 10: Find the phase angle of 1 + 1j
print("\nQuestion 10: Find the phase angle of complex number 1 + 1j")
import cmath
z10 = 1 + 1j
phase_angle = cmath.phase(z10)
print(phase_angle)
print("Phase angle in degrees:", phase_angle * 180 / 3.141592653589793)
```
